```python
import pandas as pd
import numpy as np

df = pd.read_csv("../data/loan_data_combined.csv")
```

```python
df["income_band"] = pd.qcut(
    df["income"],
    q=4,
    labels=["Low", "Medium", "High", "Very High"]
)
```

```python
df["income_band"].value_counts()
```

```python
bins = [20, 30, 45, 60, 100]
labels = ["21-30", "31-45", "46-60", "60+"]

df["age_group"] = pd.cut(
    df["age"],
    bins=bins,
    labels=labels
)
```

```python
df["age_group"].value_counts()
```

```python
def employment_stability(row):
    if row["current_job_yrs"] >= 5 and row["experience"] >= 10:
        return "Stable"
    elif row["current_job_yrs"] >= 2 and row["experience"] >= 5:
        return "Moderate"
    else:
        return "Unstable"

df["employment_stability"] = df.apply(employment_stability, axis=1)
```

```python
df["employment_stability"].value_counts()
```

```python
df_train = df[df["source"] == "train"]

pd.crosstab(
    df_train["employment_stability"],
    df_train["risk_flag"],
    normalize="index"
)
```
